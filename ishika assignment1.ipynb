```python
import random

numbers = []

for i in range(10):
    numbers.append(random.randint(1, 100))

print("Random Numbers:", numbers)

print("Even Numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "Python is easy"

print("Reverse String:")
print(text[::-1])

print("Reverse Words:")
words = text.split()
words.reverse()
print(" ".join(words))
```
